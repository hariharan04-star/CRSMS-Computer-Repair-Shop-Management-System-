```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import joblib
import json
```

```python
df = pd.read_csv("House_Price_Prediction_MLOps.csv")

X = df[['Area_sqft', 'Bedrooms', 'Bathrooms',
        'House_Age_Years', 'Distance_to_City_km', 'Price_Lakh']]
```

```python
print("Dataset Columns:", df.columns.tolist())
display(df.head())
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

X = df[['Area_sqft', 'Bedrooms', 'Bathrooms', 'House_Age_Years', 'Distance_to_City_km']]
y = df['Price_Lakh']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Training Regression Model for House Prices...")
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

predictions = model.predict(X_test)
mae = mean_absolute_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print(f"Mean Absolute Error: {mae:.2f} Lakh")
print(f"R2 Score: {r2:.4f}")

metrics = {"mae": mae, "r2_score": r2}
with open("metrics.json", "w") as f:
    json.dump(metrics, f, indent=4)

print("Training Completed Successfully")
```
